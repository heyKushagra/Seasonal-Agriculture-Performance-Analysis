# **9. Outlier Analysis**

Investigate potential outliers using statistical and visual methods.  
This is done to Decide whether an extreme value is:
* A data entry problem,
* A legitimate observation,
* Or something requiring further investigation.

```python
# IQR based Outlier Summary for numerical columns

outlier_summary = []

for col in numeric_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        "Column": col,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": count,
        "Outlier_Percentage": round(count / len(df) * 100, 2)
    })

outlier_summary = pd.DataFrame(outlier_summary).sort_values(
    "Outlier_Count", ascending=False
)

outlier_summary
```

# **Student Analysis - 1**

**Question** - Which season uses water more efficiently in terms of agricultural yield?  
**Analysis** - Compare the average Water_Efficiency_t_per_1000m3 across Kharif, Rabi and Zaid.  
**Visualization**- Bar plot of Average Water Efficiency by Season.  
**Interpretation** -
This analysis helps identify whether some seasons produce more yield for the amount of water used. It adds a resource-efficiency perspective rather than looking only at total water consumption.

```python
#Visualization
# Which season uses water more efficiently in terms of agricultural yield?

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Season",
    y="Water_Efficiency_t_per_1000m3",
    errorbar=None
)

plt.title("Average Water Efficiency Across Seasons")
plt.xlabel("Season")
plt.ylabel("Water Efficiency (tonnes/1000m³)")
plt.tight_layout()
plt.show()
```

# **Student Analysis - 2 (Fertilizer Use vs Yield)**

**Question** - Does higher fertilizer usage show any relationship with agricultural yield?  
**Analysis** - Compare Fertilizer_kg_ha with Yield_Tonnes_Ha for all farms.  
**Visualization**- Scatter plot with fertilizer usage on the X axis and yield on the Y axis.  
**Interpretation** -
This analysis can show whether farms using different amounts of fertilizer tend to have different yield levels. It is useful because fertilizer is an important input, but the relationship should be treated as an association rather than proof that fertilizer directly causes higher yield.

```python
# Visualization - Student Analysis 2
# Does higher fertilizer usage show any relationship with agricultural yield?

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="Fertilizer_kg_ha",
    y="Yield_Tonnes_Ha"
)
plt.title("Fertilizer Usage vs Agricultural Yield")
plt.xlabel("Fertilizer Used (kg/ha)")
plt.ylabel("Yield (tonnes/ha)")
plt.tight_layout()
plt.show()
```

# **Student Analysis - 3 (Return on Investment (ROI) by Season)**
**Question** - Which season provides better returns compared with the cost of farming?  
**Analysis** - Calculate the profit-to-cost ratio for each farm and compare the average ratio across seasons.  
**Visualization**- Bar plot of Average Profit-to-Cost Ratio by Season.  
**Interpretation** -
This analysis looks beyond total profit and checks how efficiently the money spent on farming translates into profit. A higher ratio means more profit relative to the cost involved.

```python
# Visualization - Student Analysis 3

# Which season provides better returns compared with the cost of farming?

df["Profit_Cost_Ratio"] = df["Profit_INR"] / df["Total_Cost_INR"]

plt.figure(figsize=(8, 5))
sns.barplot(
    data=df,
    x="Season",
    y="Profit_Cost_Ratio",
    errorbar=None
)
plt.title("Average Profit-to-Cost Ratio Across Seasons")
plt.xlabel("Season")
plt.ylabel("Profit-to-Cost Ratio")
plt.tight_layout()
plt.show()
```

# **14. Key Insights**

* Observation: The box plot shows a wide range of profit values and several points above the main distribution.
* Evidence: Profit distribution box plot.
* Interpretation: Profit is not consistent for every agricultural activity, even within the same season.

4. Zaid has comparatively lower profit distribution

* Observation: The main profit distribution for Zaid is lower than that of Kharif and Rabi.
* Evidence: Seasonal profit box plot.
* Interpretation: Zaid activities generally show lower profit values in the available data.
* Limitation: This does not mean Zaid farming is always less profitable because the dataset may contain different crops, regions, and farming conditions.

5. `Rainfall and yield show a positive pattern`

* Observation: Higher rainfall values are associated with many of the higher yield observations.
* Evidence: Rainfall vs Agricultural Yield scatter plot.
* Interpretation: Rainfall appears to have a relationship with agricultural yield in the dataset.
* Limitation: This is an association, not proof that rainfall alone causes higher yield.

6. `Yield is highly spread at similar rainfall levels`

* Observation: At similar rainfall levels, yield values vary considerably.
* Evidence: Scatter plot of rainfall against yield.
* Interpretation: Rainfall alone does not explain all the variation in agricultural yield.
* Limitation: Other factors such as soil conditions, crop type, farming practices, and resource usage are not isolated in this graph.

7. `The dataset contains unusual profit observations`

* Observation: Several profit values appear far above the main box-plot range.
* Evidence: Outliers visible in the profit box plot.
* Interpretation: A small number of observations have unusually high profits compared with the majority.
* Limitation: These observations should not automatically be treated as incorrect without checking the original records.

8. `Seasonal comparison is useful for identifying areas for further analysis`

* Observation: Both yield and profit distributions change across Kharif, Rabi, and Zaid.
* Evidence: Seasonal yield bar chart and profit box plot.
* Interpretation: Season is an important category for comparing agricultural performance in this dataset.
* Limitation: Season alone cannot explain the differences; regional, environmental, production, and economic factors should also be examined.
